```python
import shutil
import os

# --- CẤU HÌNH ---
# Tên thư mục bạn muốn nén
directory_to_zip = '/kaggle/working/coco_features_2048d'

# Tên file zip đầu ra (bao gồm đường dẫn /kaggle/working/ để đảm bảo file được lưu đúng chỗ)
output_filename = '/kaggle/working/coco_features_download' 

# --- XỬ LÝ ---
if os.path.exists(directory_to_zip):
    print(f"Tìm thấy thư mục: {directory_to_zip}")
    print("Đang tiến hành nén file... (Việc này có thể mất vài phút)")
    
    # Tạo file zip
    shutil.make_archive(output_filename, 'zip', directory_to_zip)
    
    print(f"Nén thành công!")
    print(f"File đã được lưu tại: {output_filename}.zip")
    print("--> Hãy nhìn sang cột bên phải màn hình (Sidebar), mục 'Data' -> 'Output' để tìm và tải file về.")
    
else:
    print(f"Lỗi: Không tìm thấy thư mục {directory_to_zip}. Hãy kiểm tra lại xem bước trích xuất đã chạy xong chưa.")
```
